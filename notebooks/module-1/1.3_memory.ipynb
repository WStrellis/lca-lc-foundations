## No memory

```python
from langchain.agents import create_agent
from langchain_openrouter import ChatOpenRouter

model = ChatOpenRouter(
    model="openrouter/owl-alpha",
    temperature=0.5,
)


agent = create_agent(
    model=model
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Westley and I hiked the Zion Traverse in March.")

response = agent.invoke(
    {"messages": [question]} 
)
```

```python
from pprint import pprint

pprint(response)
```

```python
question = HumanMessage(content="what trail did I hike in March?")

response = agent.invoke(
    {"messages": [question]} 
)

pprint(response)
```

## Memory

```python
from langgraph.checkpoint.memory import InMemorySaver  


agent = create_agent(
    model=model,
    checkpointer=InMemorySaver(),  
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Westley and I hiked the Zion Traverse in March.")
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke(
    {"messages": [question]},
    config=config,  
)
```

```python
pprint(response)
```

```python
question = HumanMessage(content="what trail did I hike in March?")

response = agent.invoke(
    {"messages": [question]},
    config=config,  
)

pprint(response.messages[-1].content)
```
